Objective and Introduction
When looking for a product, we are scared to purchase the cheapest but unwilling to have the most expensive item. 
Neither is the one with the highest sales now the lowest. Result in spending redundant time on looking for a satisfying product on our own. 
The customers’ behavior, choosing the cheapest product, does not align with the traditional demand and supply curve, which inspires our group to make a hypothesis that,
Price is not the critical factor affecting  the Sales in real life.

Our project researched on the factors affecting may affect the sales of products on amazon.com, including price per count, comment count and star rating.
By referencing the price, sales, and # of comments data of Coffee Pod, Sanitary Pad, Induction Cook  Top on Amazon with zip code 10001 (New York) by crawling and clustering.
We want to identify the most relevant factor that is impacting its sales instead of price per count.
